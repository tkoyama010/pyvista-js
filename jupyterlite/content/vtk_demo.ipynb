# PyVista-js with vtk.js Rendering

Interactive 3D visualization using vtk.js in JupyterLite.

```python
# Load vtk.js library
from IPython.display import display, HTML

display(HTML('''
<script src="https://unpkg.com/vtk.js@29.5.0"></script>
<script>
  console.log('vtk.js version:', vtk.version);
</script>
'''))

print('✅ vtk.js loaded!')
```

```python
# Wait for vtk.js to load
import time
time.sleep(1)
print('Ready!')
```

```python
# Import JavaScript objects
import js

# Create rendering container
display(HTML('<div id="vtk-container" style="width:600px;height:400px;border:1px solid #ccc;"></div>'))
```

```python
# Initialize vtk.js renderer
import time
time.sleep(0.2)  # Wait for DOM

container = js.document.getElementById('vtk-container')
js.console.log('Container:', container)

# Create renderer
renderer = js.vtk.Rendering.Core.vtkRenderer.newInstance()
renderer.setBackground(0.2, 0.3, 0.4)

# Create render window
render_window = js.vtk.Rendering.Core.vtkRenderWindow.newInstance()
render_window.addRenderer(renderer)

# Create OpenGL render window
opengl_rw = js.vtk.Rendering.OpenGL.vtkRenderWindow.newInstance()
opengl_rw.setContainer(container)
opengl_rw.setSize(600, 400)
render_window.addView(opengl_rw)

# Create interactor
interactor = js.vtk.Rendering.Core.vtkRenderWindowInteractor.newInstance()
interactor.setView(render_window)
interactor.initialize()
interactor.bindEvents(container)
interactor.setInteractorStyle(js.vtk.Interaction.Style.vtkInteractorStyleTrackballCamera.newInstance())

print('✅ vtk.js renderer initialized!')
```

```python
# Create a red sphere
sphere_source = js.vtk.Filters.Sources.vtkSphereSource.newInstance()
sphere_source.setCenter(0, 0, 0)
sphere_source.setRadius(1.0)
sphere_source.setThetaResolution(30)
sphere_source.setPhiResolution(30)

# Create mapper
mapper = js.vtk.Rendering.Core.vtkMapper.newInstance()
mapper.setInputConnection(sphere_source.getOutputPort())

# Create actor
actor = js.vtk.Rendering.Core.vtkActor.newInstance()
actor.setMapper(mapper)
actor.getProperty().setColor(1.0, 0.0, 0.0)
actor.getProperty().setOpacity(0.8)

# Add to renderer
renderer.addActor(actor)

print('✅ Sphere added!')
```

```python
# Render the scene
renderer.resetCamera()
render_window.render()

print('✅ Scene rendered! You should see a red sphere above.')
print('   Use your mouse to rotate/zoom the view.')
```
