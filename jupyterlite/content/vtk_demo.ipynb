# PyVista-js with vtk.js Rendering

Interactive 3D visualization using vtk.js in JupyterLite.

```python
# Load vtk.js library
from IPython.display import display, HTML

display(HTML('''
<script src="https://unpkg.com/vtk.js@29.5.0"></script>
<script>
  console.log('vtk.js loaded, version:', vtk.version);
</script>
'''))

print('✅ vtk.js loading...')
```

```python
# Wait for vtk.js to load
import time
time.sleep(2)
print('Ready!')
```

```python
# Create rendering container and initialize vtk.js in one script
display(HTML('''
<div id="vtk-container" style="width:600px;height:400px;border:1px solid #ccc;"></div>
<script>
(function() {
  // Wait a moment for container to be ready
  setTimeout(function() {
    const container = document.getElementById('vtk-container');
    
    // Create renderer
    const renderer = vtk.Rendering.Core.vtkRenderer.newInstance();
    renderer.setBackground(0.2, 0.3, 0.4);
    
    // Create render window
    const renderWindow = vtk.Rendering.Core.vtkRenderWindow.newInstance();
    renderWindow.addRenderer(renderer);
    
    // Create OpenGL render window
    const openGLRenderWindow = vtk.Rendering.OpenGL.vtkRenderWindow.newInstance();
    openGLRenderWindow.setContainer(container);
    openGLRenderWindow.setSize(600, 400);
    renderWindow.addView(openGLRenderWindow);
    
    // Create interactor
    const interactor = vtk.Rendering.Core.vtkRenderWindowInteractor.newInstance();
    interactor.setView(renderWindow);
    interactor.initialize();
    interactor.bindEvents(container);
    interactor.setInteractorStyle(
      vtk.Interaction.Style.vtkInteractorStyleTrackballCamera.newInstance()
    );
    
    // Create a red sphere
    const sphereSource = vtk.Filters.Sources.vtkSphereSource.newInstance();
    sphereSource.setCenter(0, 0, 0);
    sphereSource.setRadius(1.0);
    sphereSource.setThetaResolution(30);
    sphereSource.setPhiResolution(30);
    
    // Create mapper
    const mapper = vtk.Rendering.Core.vtkMapper.newInstance();
    mapper.setInputConnection(sphereSource.getOutputPort());
    
    // Create actor
    const actor = vtk.Rendering.Core.vtkActor.newInstance();
    actor.setMapper(mapper);
    actor.getProperty().setColor(1.0, 0.0, 0.0);
    actor.getProperty().setOpacity(0.8);
    
    // Add to renderer
    renderer.addActor(actor);
    
    // Render
    renderer.resetCamera();
    renderWindow.render();
    
    console.log('✅ vtk.js scene rendered!');
    
    // Store globally for Python access
    window.vtkScene = {
      renderer: renderer,
      renderWindow: renderWindow,
      interactor: interactor
    };
  }, 300);
})();
</script>
'''))

print('✅ Scene created!')
print('   A red sphere should appear above.')
print('   Use your mouse to rotate and zoom!')
```
