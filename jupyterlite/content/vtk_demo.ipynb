# PyVista-js with vtk.js Rendering

Interactive 3D visualization using vtk.js in JupyterLite.

```python
# Load vtk.js library
from IPython.display import display, HTML

display(HTML('''
<script src="https://unpkg.com/vtk.js@29.5.0"></script>
<script>
  setTimeout(function() {
    if (typeof vtk !== 'undefined') {
      console.log('✅ vtk.js loaded successfully, version:', vtk.version);
    } else {
      console.error('❌ vtk.js failed to load');
    }
  }, 1000);
</script>
'''))

print('✅ Loading vtk.js...')
```

```python
# Wait for vtk.js to load
import time
time.sleep(2)
print('✅ Ready to render!')
```

```python
# Create rendering container and initialize vtk.js
display(HTML('''
<div id="vtk-container" style="width:600px;height:400px;border:2px solid #333;background:#1a1a1a;"></div>
<div id="debug-output" style="margin-top:10px;padding:10px;background:#f0f0f0;font-family:monospace;font-size:12px;"></div>
<script>
(function() {
  const debugEl = document.getElementById('debug-output');
  function log(msg) {
    console.log(msg);
    debugEl.innerHTML += msg + '<br>';
  }
  
  setTimeout(function() {
    try {
      log('🔍 Starting vtk.js initialization...');
      
      const container = document.getElementById('vtk-container');
      if (!container) {
        log('❌ Container not found');
        return;
      }
      log('✅ Container found: ' + container.clientWidth + 'x' + container.clientHeight);
      
      if (typeof vtk === 'undefined') {
        log('❌ vtk.js not loaded');
        return;
      }
      log('✅ vtk.js available');
      
      // Create renderer
      const renderer = vtk.Rendering.Core.vtkRenderer.newInstance();
      renderer.setBackground(0.2, 0.3, 0.4);
      log('✅ Renderer created');
      
      // Create render window
      const renderWindow = vtk.Rendering.Core.vtkRenderWindow.newInstance();
      renderWindow.addRenderer(renderer);
      log('✅ Render window created');
      
      // Create OpenGL render window
      const openGLRenderWindow = vtk.Rendering.OpenGL.vtkRenderWindow.newInstance();
      openGLRenderWindow.setContainer(container);
      openGLRenderWindow.setSize(600, 400);
      renderWindow.addView(openGLRenderWindow);
      log('✅ OpenGL window created');
      
      // Create interactor
      const interactor = vtk.Rendering.Core.vtkRenderWindowInteractor.newInstance();
      interactor.setView(renderWindow);
      interactor.initialize();
      interactor.bindEvents(container);
      interactor.setInteractorStyle(
        vtk.Interaction.Style.vtkInteractorStyleTrackballCamera.newInstance()
      );
      log('✅ Interactor created');
      
      // Create a red sphere
      const sphereSource = vtk.Filters.Sources.vtkSphereSource.newInstance();
      sphereSource.setCenter(0, 0, 0);
      sphereSource.setRadius(1.0);
      sphereSource.setThetaResolution(30);
      sphereSource.setPhiResolution(30);
      log('✅ Sphere source created');
      
      // Create mapper
      const mapper = vtk.Rendering.Core.vtkMapper.newInstance();
      mapper.setInputConnection(sphereSource.getOutputPort());
      log('✅ Mapper created');
      
      // Create actor
      const actor = vtk.Rendering.Core.vtkActor.newInstance();
      actor.setMapper(mapper);
      actor.getProperty().setColor(1.0, 0.0, 0.0);
      actor.getProperty().setOpacity(0.8);
      log('✅ Actor created');
      
      // Add to renderer
      renderer.addActor(actor);
      log('✅ Actor added to renderer');
      
      // Render
      renderer.resetCamera();
      renderWindow.render();
      log('🎉 Scene rendered successfully!');
      log('👆 Red sphere should be visible above. Try dragging to rotate!');
      
    } catch (error) {
      log('❌ Error: ' + error.message);
      console.error(error);
    }
  }, 500);
})();
</script>
'''))

print('Check the debug output below the render area!')
```
