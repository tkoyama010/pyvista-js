Basic usage:


```python
import pyvista_js as pv
plotter = pv.Plotter()
_ = plotter.add_mesh(pv.Sphere())
plotter.add_axes()
plotter.show()  # doctest: +SKIP
```

With multiple meshes:


```python
import pyvista_js as pv
plotter = pv.Plotter()
_ = plotter.add_mesh(pv.Sphere(), color='red')
_ = plotter.add_mesh(pv.Cube(center=(2, 0, 0)), color='blue')
plotter.add_axes()
plotter.show()  # doctest: +SKIP
```
