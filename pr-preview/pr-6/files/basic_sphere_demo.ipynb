# Basic Example: Creating and Displaying a Sphere

This notebook demonstrates how to create and display a 3D sphere using pyvista-js.

```python
# Install pyvista-js from wheel file
import micropip
await micropip.install('numpy')

# Get the base URL dynamically
import js
origin = js.location.origin
pathname = js.location.pathname
base_url = str(origin) + str(pathname).rsplit('/', 2)[0]
wheel_url = f"{base_url}/files/pyvista_js-0.1.dev0-py3-none-any.whl"
print(f"Installing from: {wheel_url}")
await micropip.install(wheel_url)
```

```python
import pyvista_js as pv
```

```python
# Create a plotter
plotter = pv.Plotter()
```

```python
# Create a sphere
sphere = pv.Sphere(radius=1.0, center=(0, 0, 0))

print(f"Sphere has {sphere.n_points} points")
```

```python
# Add the sphere to the plotter
plotter.add_mesh(sphere, color='red', opacity=0.8)
```

```python
# Display the visualization
plotter.show()
```
