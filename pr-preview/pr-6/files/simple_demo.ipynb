# PyVista-js Simple Demo

A simple demonstration with inline code (no external imports needed).

```python
import micropip
await micropip.install('numpy')
```

```python
import numpy as np

class Sphere:
    def __init__(self, radius=1.0, center=(0, 0, 0)):
        self.radius = radius
        self.center = np.array(center)
        theta = np.linspace(0, 2*np.pi, 30)
        phi = np.linspace(0, np.pi, 30)
        points = []
        for p in phi:
            for t in theta:
                x = radius * np.sin(p) * np.cos(t) + center[0]
                y = radius * np.sin(p) * np.sin(t) + center[1]
                z = radius * np.cos(p) + center[2]
                points.append([x, y, z])
        self.points = np.array(points)
        self.n_points = len(points)

class Plotter:
    def __init__(self):
        self.meshes = []
    
    def add_mesh(self, mesh, color='red', opacity=0.8):
        self.meshes.append({'mesh': mesh, 'color': color, 'opacity': opacity})
        print(f"Added {mesh.__class__.__name__} with {mesh.n_points} points")
    
    def show(self):
        print(f"\nScene contains {len(self.meshes)} mesh(es)")
        for i, item in enumerate(self.meshes):
            print(f"  Mesh {i+1}: {item['mesh'].n_points} points, color={item['color']}")

print('✅ Classes loaded!')
```

```python
plotter = Plotter()
```

```python
sphere = Sphere(radius=1.0, center=(0, 0, 0))
print(f"Sphere has {sphere.n_points} points")
```

```python
plotter.add_mesh(sphere, color='red', opacity=0.8)
```

```python
plotter.show()
```
