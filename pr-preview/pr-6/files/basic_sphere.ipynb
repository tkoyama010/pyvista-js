# Basic Example: Creating and Displaying a Sphere

This notebook demonstrates how to create and display a 3D sphere using pyvista-js.

```python
import pyvista_js as pv
```

```python
# Create a plotter
plotter = pv.Plotter()
```

```python
# Create a sphere
sphere = pv.Sphere(radius=1.0, center=(0, 0, 0))

print(f"Sphere has {sphere.n_points} points")
```

```python
# Add the sphere to the plotter
plotter.add_mesh(sphere, color='red', opacity=0.8)
```

```python
# Display the visualization
plotter.show()
```
